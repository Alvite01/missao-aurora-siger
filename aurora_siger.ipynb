## 5.2 Algoritmo de verificação

A função abaixo é o algoritmo em código: testa as condições **em sequência**, na
mesma ordem do fluxograma — integridade estrutural → energia → temperaturas →
pressão dos tanques → módulos críticos. A primeira condição que falhar já retorna
"DECOLAGEM ABORTADA" com o motivo específico; só chega a "PRONTO PARA DECOLAR"
quem passar por todas as verificações.

> **Sobre os módulos críticos:** o `for` no final da função percorre **todos** os
> módulos (propulsão, navegação, suporte de vida, comunicação) e aborta se
> qualquer um deles não estiver "online" — não é só um caso isolado. Testamos
> isso na seção 5.3 rodando o algoritmo contra os 10 cenários de teste definidos
> na seção 5.1, incluindo cenários com falha em módulos diferentes, e o
> resultado bateu em todos eles.

No relatório em PDF, este mesmo algoritmo entra como fluxograma (imagem) — o
código abaixo é a tradução dele pra Python.


```python
def verificar_decolagem(dados, faixas):
    if dados["integridade_estrutural"] != 1:
        return "DECOLAGEM ABORTADA", "estrutura comprometida"

    if dados["energia_percentual"] < faixas["energia_percentual_min"]:
        return "DECOLAGEM ABORTADA", "energia insuficiente"

    temp_int_min, temp_int_max = faixas["temperatura_interna_c"]
    if not (temp_int_min <= dados["temperatura_interna_c"] <= temp_int_max):
        return "DECOLAGEM ABORTADA", "temperatura interna fora da faixa"

    temp_ext_min, temp_ext_max = faixas["temperatura_externa_c"]
    if not (temp_ext_min <= dados["temperatura_externa_c"] <= temp_ext_max):
        return "DECOLAGEM ABORTADA", "temperatura externa fora da faixa"

    p_comb_min, p_comb_max = faixas["pressao_combustivel_bar"]
    if not (p_comb_min <= dados["pressao_combustivel_bar"] <= p_comb_max):
        return "DECOLAGEM ABORTADA", "pressão do combustível anômala"

    p_oxid_min, p_oxid_max = faixas["pressao_oxidante_bar"]
    if not (p_oxid_min <= dados["pressao_oxidante_bar"] <= p_oxid_max):
        return "DECOLAGEM ABORTADA", "pressão do oxidante anômala"

    # percorre TODOS os módulos críticos (não só o primeiro) e aborta se
    # qualquer um deles não estiver "online"
    for modulo, status in dados["modulos_criticos"].items():
        if status != "online":
            return "DECOLAGEM ABORTADA", f"módulo crítico offline: {modulo}"

    return "PRONTO PARA DECOLAR", "todas as verificações passaram"
```

```python
# 1) leitura dos dados -> já temos em `telemetria` (cenário ativo, definido na seção 5.1)
# 2) execução das verificações
status, motivo = verificar_decolagem(telemetria, faixas_seguras)

# 3) resultado final impresso
print(f"STATUS: {status}")
print(f"MOTIVO: {motivo}")

# 4) verificação extra: roda o mesmo algoritmo contra TODOS os cenários de teste
#    da seção 5.1, pra demonstrar que ele lida corretamente com o caso de sucesso
#    e com cada tipo de falha (estrutura, energia, temperatura, pressão, módulo)
print()
print("Resultado para todos os cenários de teste:")
for nome, dados in telemetrias_teste.items():
    status_teste, motivo_teste = verificar_decolagem(dados, faixas_seguras)
    print(f"  {nome}: {status_teste} — {motivo_teste}")
```

## 5.4 Análise energética

Calculamos a autonomia da nave a partir de quatro números que o enunciado pede
explicitamente: capacidade total da bateria, carga atual, perdas energéticas e
consumo estimado durante a decolagem.

**A conta**: primeiro descobrimos quanta energia (em kWh) está de fato disponível
— capacidade × carga atual, menos as perdas. Depois dividimos isso pelo consumo
por hora, pra saber por quantas horas a nave aguenta operar nesse ritmo.

Os quatro valores abaixo (capacidade total, perdas, consumo estimado na
decolagem) são os valores definitivos adotados pelo grupo.


```python
capacidade_total_kwh = 60          # capacidade máxima da bateria
carga_atual_percent = telemetria["energia_percentual"]
perdas_kwh = 2.5                   # perdas térmicas e de conversão estimadas
consumo_decolagem_kwh_h = 18       # consumo previsto durante a decolagem

energia_disponivel_kwh = (capacidade_total_kwh * carga_atual_percent / 100) - perdas_kwh
autonomia_horas = energia_disponivel_kwh / consumo_decolagem_kwh_h

print(f"Energia disponível: {energia_disponivel_kwh:.2f} kWh")
print(f"Autonomia estimada: {autonomia_horas:.2f} horas")
```
